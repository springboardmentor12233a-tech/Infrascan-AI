```python
import os
import shutil
import random

source = "/content/InfraScan_Dataset"
output = "/content/InfraScan_Split"

random.seed(42)

for split in ["train", "validation"]:
    for category in ["cracked", "uncracked"]:
        os.makedirs(
            os.path.join(output, split, category),
            exist_ok=True
        )

for category in ["cracked", "uncracked"]:
    files = [
        f for f in os.listdir(os.path.join(source, category))
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(files)
    split_index = max(1, int(len(files) * 0.8))

    for filename in files[:split_index]:
        shutil.copy2(
            os.path.join(source, category, filename),
            os.path.join(output, "train", category, filename)
        )

    for filename in files[split_index:]:
        shutil.copy2(
            os.path.join(source, category, filename),
            os.path.join(output, "validation", category, filename)
        )

print("Sample dataset split completed!")

for split in ["train", "validation"]:
    for category in ["cracked", "uncracked"]:
        folder = os.path.join(output, split, category)
        print(f"{split.title()} - {category}: {len(os.listdir(folder))}")
```

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(256, 256, 3)),
    tf.keras.layers.Rescaling(1.0 / 255),

    tf.keras.layers.Conv2D(16, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import tensorflow as tf

train_dataset = tf.keras.utils.image_dataset_from_directory(
    "/content/InfraScan_Split/train",
    image_size=(256, 256),
    batch_size=4,
    label_mode="binary",
    shuffle=True,
    seed=42
)

validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "/content/InfraScan_Split/validation",
    image_size=(256, 256),
    batch_size=4,
    label_mode="binary",
    shuffle=False
)

print("Training and validation datasets loaded!")
print("Classes:", train_dataset.class_names)
```

```python
history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=5
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))

plt.plot(history.history["accuracy"], marker="o", label="Training Accuracy")
plt.plot(history.history["val_accuracy"], marker="o", label="Validation Accuracy")

plt.title("InfraScan CNN Training Progress")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.xticks(range(5), range(1, 6))
plt.ylim(0, 1)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "/content/InfraScan_Training_Accuracy.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

print("Training accuracy graph saved!")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred = []

for images, labels in validation_dataset:
    scores = model.predict(images, verbose=0).flatten()
    predictions = (scores < 0.5).astype(int)

    y_true.extend(labels.numpy().flatten().astype(int))
    y_pred.extend(predictions)

print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred, labels=[0, 1]))

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=validation_dataset.class_names,
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=validation_dataset.class_names,
    zero_division=0
)

matrix = confusion_matrix(y_true, y_pred, labels=[0, 1])

with open("/content/InfraScan_Evaluation_Report.txt", "w") as f:
    f.write("INFRASCAN - PRELIMINARY MODEL EVALUATION\n\n")
    f.write("Dataset: 14-image test sample\n")
    f.write("Validation images: 3\n\n")
    f.write("Confusion Matrix:\n")
    f.write(str(matrix))
    f.write("\n\nClassification Report:\n")
    f.write(report)
    f.write("\n\nNote: Results are preliminary and not reliable\n")
    f.write("for real-world crack detection.\n")

print("Evaluation report saved successfully!")
```

```python
import os
import numpy as np
from PIL import Image

root = "/content/InfraScan_Preprocessed"
total = 0
errors = []

for category in sorted(os.listdir(root)):
    folder = os.path.join(root, category)
    if not os.path.isdir(folder):
        continue

    count = 0
    for filename in os.listdir(folder):
        path = os.path.join(folder, filename)
        try:
            with Image.open(path) as img:
                img = img.convert("RGB").resize((256, 256))
                pixels = np.asarray(img, dtype=np.float32) / 255.0
                assert pixels.shape == (256, 256, 3)
                count += 1
        except Exception as e:
            errors.append((filename, str(e)))

    print(f"{category}: {count} images verified and preprocessed")
    total += count

print("\nTotal images:", total)
print("Errors:", len(errors))
print("Pixel normalization range: 0 to 1")
print("Preprocessing verification completed.")
```

```python
import tensorflow as tf

data_dir = "/content/InfraScan_Preprocessed"

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(256, 256),
    batch_size=16
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(256, 256),
    batch_size=16
)

class_names = train_ds.class_names
print("Classes:", class_names)

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(256, 256, 3)),
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(len(class_names), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)

model.save("/content/InfraScan_6Class_CNN.keras")
print("Six-class CNN trained and saved.")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred = []

for images, labels in val_ds:
    predictions = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(predictions, axis=1))

print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(len(class_names))),
    target_names=class_names,
    zero_division=0
))
```

```python

# ============================================================
# INFRASCAN — COMPLETE DEMO APPLICATION
# Login | Role Access | Asset Management | Image Inspection
# ============================================================

import os
import sys
import sqlite3
import hashlib
import subprocess
import importlib.util

# 1. Install Gradio if missing
if importlib.util.find_spec("gradio") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "gradio"
    ])

import gradio as gr
import numpy as np
from PIL import Image

DB_PATH = "/content/InfraScan.db"
MODEL_PATH = "/content/InfraScan_6Class_CNN.keras"

# 2. Database initialization
def connect_db():
    return sqlite3.connect(DB_PATH, timeout=30)

def hash_password(password):
    return hashlib.sha256(password.encode("utf-8")).hexdigest()

with connect_db() as conn:
    conn.execute("""
        CREATE TABLE IF NOT EXISTS users (
            username TEXT PRIMARY KEY,
            password_hash TEXT NOT NULL,
            role TEXT NOT NULL
        )
    """)
    conn.execute("""
        CREATE TABLE IF NOT EXISTS assets (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            asset_name TEXT NOT NULL,
            asset_type TEXT NOT NULL,
            location TEXT,
            status TEXT DEFAULT 'Active'
        )
    """)

    conn.execute(
        "INSERT OR IGNORE INTO users VALUES (?, ?, ?)",
        ("admin", hash_password("Admin@123"), "admin")
    )
    conn.execute(
        "INSERT OR IGNORE INTO users VALUES (?, ?, ?)",
        ("user", hash_password("User@123"), "user")
    )

    conn.execute("""
        INSERT INTO assets (asset_name, asset_type, location, status)
        SELECT 'Bridge A', 'Bridge', 'Mysore', 'Active'
        WHERE NOT EXISTS (
            SELECT 1 FROM assets WHERE asset_name = 'Bridge A'
        )
    """)

print("Database ready:", DB_PATH)

# 3. Load optional CNN model
model = None
class_names = [
    "deck_cracked",
    "deck_uncracked",
    "pavement_cracked",
    "pavement_uncracked",
    "wall_cracked",
    "wall_uncracked"
]

if os.path.isfile(MODEL_PATH):
    try:
        import tensorflow as tf
        model = tf.keras.models.load_model(MODEL_PATH)
        print("CNN model loaded successfully.")
    except Exception as e:
        print("CNN model unavailable:", str(e))
else:
    print("CNN model not found; image preprocessing remains available.")

# 4. Login
def login(username, password):
    username = (username or "").strip()

    with connect_db() as conn:
        row = conn.execute(
            "SELECT password_hash, role FROM users WHERE username = ?",
            (username,)
        ).fetchone()

    if row and row[0] == hash_password(password or ""):
        role = row[1]
        return (
            f"Login successful: {username} ({role})",
            role,
            gr.update(visible=True),
            gr.update(visible=(role == "admin"))
        )

    return (
        "Invalid username or password.",
        None,
        gr.update(visible=False),
        gr.update(visible=False)
    )

def logout():
    return (
        None,
        "Logged out.",
        gr.update(visible=False),
        gr.update(visible=False)
    )

# 5. Register infrastructure asset
def register_asset(name, asset_type, location, role):
    if role != "admin":
        return "Access denied. Admin login required."

    name = (name or "").strip()
    location = (location or "").strip()

    if not name or not asset_type:
        return "Please enter the asset name and type."

    try:
        with connect_db() as conn:
            conn.execute("""
                INSERT INTO assets
                    (asset_name, asset_type, location, status)
                VALUES (?, ?, ?, 'Active')
            """, (name, asset_type, location))

        return f"Asset '{name}' registered successfully."

    except Exception as e:
        return f"Asset registration failed: {e}"

# 6. View assets — returns plain text
def view_assets(role):
    if role not in ("admin", "user"):
        return "Please log in first."

    try:
        with connect_db() as conn:
            rows = conn.execute("""
                SELECT id, asset_name, asset_type, location, status
                FROM assets
                ORDER BY id DESC
            """).fetchall()

        if not rows:
            return "No assets registered yet."

        lines = [
            "ID | Asset Name | Type | Location | Status",
            "-" * 58
        ]

        for row in rows:
            lines.append(
                f"{row[0]} | {row[1]} | {row[2]} | "
                f"{row[3] or '-'} | {row[4]}"
            )

        return "\n".join(lines)

    except Exception as e:
        return f"Could not load assets: {e}"

# 7. Image preprocessing and experimental prediction
def inspect_image(image, role):
    if role not in ("admin", "user"):
        return None, "Please log in first."

    if image is None:
        return None, "Please upload a concrete image."

    try:
        img = Image.fromarray(np.asarray(image)).convert("RGB")
        img = img.resize((256, 256))
        arr = np.asarray(img, dtype=np.float32) / 255.0

        if model is None:
            return img, (
                "Preprocessing successful.\n"
                "Image size: 256 x 256\n"
                "Color: RGB\n"
                "Normalized pixel range: 0–1\n"
                "Prediction unavailable because the CNN model is not loaded."
            )

        prediction = np.asarray(
            model.predict(np.expand_dims(arr, axis=0), verbose=0)
        ).reshape(-1)

        if len(prediction) != len(class_names):
            return img, (
                f"Model returned {len(prediction)} output values. "
                "Expected six classes; check the loaded model."
            )

        idx = int(np.argmax(prediction))
        label = class_names[idx]
        confidence = float(prediction[idx]) * 100

        return img, (
            f"Predicted category: {label}\n"
            f"Model score: {confidence:.2f}%\n\n"
            "Research/demo result only. This experimental model is not "
            "reliable enough for real infrastructure safety decisions."
        )

    except Exception as e:
        return None, f"Image inspection failed: {e}"

# 8. Build Gradio interface
with gr.Blocks(title="InfraScan") as demo:
    gr.Markdown(
        "# InfraScan\n"
        "### AI-Powered Crack & Damage Assessment for Civil Infrastructure"
    )

    role_state = gr.State(value=None)

    with gr.Group():
        gr.Markdown("### Login")
        username_box = gr.Textbox(label="Username", value="admin")
        password_box = gr.Textbox(
            label="Password", type="password", value="Admin@123"
        )
        login_button = gr.Button("Login", variant="primary")
        login_status = gr.Textbox(label="Login status", interactive=False)
        logout_button = gr.Button("Logout")

    with gr.Group(visible=False) as main_area:
        with gr.Tabs():
            with gr.Tab("Inspect Image"):
                gr.Markdown("Upload a concrete image for analysis.")
                image_input = gr.Image(
                    type="numpy", label="Upload concrete image"
                )
                inspect_button = gr.Button("Inspect Image")
                processed_output = gr.Image(label="Processed image")
                prediction_output = gr.Textbox(
                    label="Inspection result", lines=5
                )

            with gr.Tab("Register Asset", visible=False) as admin_tab:
                gr.Markdown("Admin-only asset registration")
                asset_name_box = gr.Textbox(label="Asset name")
                asset_type_box = gr.Dropdown(
                    choices=["Bridge", "Road", "Building", "Wall", "Other"],
                    label="Asset type",
                    value="Road"
                )
                asset_location_box = gr.Textbox(label="Location")
                register_button = gr.Button("Register Asset")
                register_status = gr.Textbox(
                    label="Registration status", interactive=False
                )

            with gr.Tab("View Assets"):
                refresh_button = gr.Button("Refresh asset list")
                asset_list = gr.Textbox(
                    label="Registered infrastructure assets",
                    lines=12,
                    interactive=False
                )

    login_button.click(
        fn=login,
        inputs=[username_box, password_box],
        outputs=[login_status, role_state, main_area, admin_tab]
    )

    logout_button.click(
        fn=logout,
        inputs=[],
        outputs=[role_state, login_status, main_area, admin_tab]
    )

    register_button.click(
        fn=register_asset,
        inputs=[
            asset_name_box, asset_type_box,
            asset_location_box, role_state
        ],
        outputs=[register_status]
    )

    refresh_button.click(
        fn=view_assets,
        inputs=[role_state],
        outputs=[asset_list]
    )

    inspect_button.click(
        fn=inspect_image,
        inputs=[image_input, role_state],
        outputs=[processed_output, prediction_output]
    )

print("Starting InfraScan demo...")
demo.launch(share=True)
```

```python
import tensorflow as tf

DATA_DIR = "/content/InfraScan_Split"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/val",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

class_names = train_ds.class_names

print("Classes:", class_names)
print("Dataset loaded successfully!")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Load MobileNetV2 without its original classification layer
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers
base_model.trainable = False

# Build our six-class classifier
model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),
    layers.Rescaling(1./127.5, offset=-1),
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.2),
    layers.Dense(6, activation="softmax")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10
)
```

```python
test_loss, test_accuracy = model.evaluate(test_ds)

print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
print("Test Loss:", round(test_loss, 4))
```

```python
import numpy as np
from PIL import Image
import os

base_path = "/content/DeepCrack_Dataset"

IMG_SIZE = 256

def load_dataset(image_folder, mask_folder):

    images = []
    masks = []

    image_path = os.path.join(base_path, image_folder)
    mask_path = os.path.join(base_path, mask_folder)

    image_files = sorted(os.listdir(image_path))
    mask_lookup = {
        os.path.splitext(f)[0]: f
        for f in os.listdir(mask_path)
    }

    for filename in image_files:

        name = os.path.splitext(filename)[0]

        if name not in mask_lookup:
            continue

        img = Image.open(
            os.path.join(image_path, filename)
        ).convert("RGB")

        mask = Image.open(
            os.path.join(mask_path, mask_lookup[name])
        ).convert("L")

        img = img.resize(
            (IMG_SIZE, IMG_SIZE),
            Image.Resampling.BILINEAR
        )

        mask = mask.resize(
            (IMG_SIZE, IMG_SIZE),
            Image.Resampling.NEAREST
        )

        images.append(np.array(img) / 255.0)

        mask = np.array(mask)
        masks.append((mask > 127).astype(np.float32))

    return np.array(images, dtype=np.float32), \
           np.array(masks, dtype=np.float32)[..., np.newaxis]


# Load training dataset
X_train, Y_train = load_dataset("train_img", "train_lab")

# Load testing dataset
X_test, Y_test = load_dataset("test_img", "test_lab")

print("Dataset preprocessing completed!")
print("Training images:", X_train.shape)
print("Training masks:", Y_train.shape)
print("Testing images:", X_test.shape)
print("Testing masks:", Y_test.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_split, X_val, Y_train_split, Y_val = train_test_split(
    X_train,
    Y_train,
    test_size=0.2,
    random_state=42
)

print("Training images:", X_train_split.shape)
print("Training masks:", Y_train_split.shape)

print("Validation images:", X_val.shape)
print("Validation masks:", Y_val.shape)

print("Testing images:", X_test.shape)
print("Testing masks:", Y_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# U-Net model
def build_unet(input_shape=(256, 256, 3)):

    inputs = layers.Input(input_shape)

    # Encoder block
    def encoder_block(x, filters):
        x = layers.Conv2D(filters, 3, padding="same",
                          activation="relu")(x)
        x = layers.BatchNormalization()(x)
        x = layers.Conv2D(filters, 3, padding="same",
                          activation="relu")(x)
        x = layers.BatchNormalization()(x)
        p = layers.MaxPooling2D((2, 2))(x)
        return x, p

    # Decoder block
    def decoder_block(x, skip, filters):
        x = layers.Conv2DTranspose(
            filters, 2, strides=2, padding="same"
        )(x)

        x = layers.Concatenate()([x, skip])

        x = layers.Conv2D(filters, 3, padding="same",
                          activation="relu")(x)
        x = layers.BatchNormalization()(x)
        x = layers.Conv2D(filters, 3, padding="same",
                          activation="relu")(x)
        x = layers.BatchNormalization()(x)

        return x

    # Encoder
    s1, p1 = encoder_block(inputs, 32)
    s2, p2 = encoder_block(p1, 64)
    s3, p3 = encoder_block(p2, 128)
    s4, p4 = encoder_block(p3, 256)

    # Bottleneck
    b1 = layers.Conv2D(512, 3, padding="same",
                       activation="relu")(p4)
    b1 = layers.BatchNormalization()(b1)
    b1 = layers.Conv2D(512, 3, padding="same",
                       activation="relu")(b1)
    b1 = layers.BatchNormalization()(b1)

    # Decoder
    d1 = decoder_block(b1, s4, 256)
    d2 = decoder_block(d1, s3, 128)
    d3 = decoder_block(d2, s2, 64)
    d4 = decoder_block(d3, s1, 32)

    # Output segmentation mask
    outputs = layers.Conv2D(
        1, 1, activation="sigmoid"
    )(d4)

    model = models.Model(inputs, outputs)

    return model


# Create model
unet_model = build_unet()

# Display model architecture
unet_model.summary()

print("U-Net model created successfully!")
```

```python
import tensorflow as tf

# Dice coefficient
def dice_coefficient(y_true, y_pred, smooth=1e-6):
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)

    y_true_f = tf.reshape(y_true, [-1])
    y_pred_f = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true_f * y_pred_f)

    dice = (2.0 * intersection + smooth) / (
        tf.reduce_sum(y_true_f) +
        tf.reduce_sum(y_pred_f) + smooth
    )

    return dice


# Compile U-Net
unet_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        dice_coefficient
    ]
)

print("U-Net model compiled successfully!")
```

```python
import tensorflow as tf

def dice_loss(y_true, y_pred, smooth=1e-6):

    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)

    y_true_f = tf.reshape(y_true, [-1])
    y_pred_f = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true_f * y_pred_f)

    dice = (2.0 * intersection + smooth) / (
        tf.reduce_sum(y_true_f) +
        tf.reduce_sum(y_pred_f) + smooth
    )

    return 1.0 - dice


def combined_loss(y_true, y_pred):

    bce = tf.keras.losses.binary_crossentropy(
        y_true, y_pred
    )

    d_loss = dice_loss(y_true, y_pred)

    return bce + d_loss


print("Combined BCE + Dice Loss defined successfully!")
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf
from PIL import Image

IMAGE_DIR = "/content/all_test_images/all_test_images"
MASK_DIR = "/content/all_test_labels/all_test_labels"

image_files = sorted([
    f for f in os.listdir(IMAGE_DIR)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
])

mask_files = {
    os.path.splitext(f)[0]: f
    for f in os.listdir(MASK_DIR)
    if f.lower().endswith((".png", ".jpg", ".jpeg"))
}

results = []
batch_size = 4
batch_images = []
batch_masks = []
batch_names = []

def evaluate_batch(images, masks, names):
    predictions = model(
        tf.convert_to_tensor(np.array(images)),
        training=False
    ).numpy()

    for i, prediction in enumerate(predictions):
        pred_mask = prediction[:, :, 0] > 0.5
        actual_mask = masks[i] > 127

        intersection = np.logical_and(
            pred_mask, actual_mask
        ).sum()

        union = np.logical_or(
            pred_mask, actual_mask
        ).sum()

        dice = (
            2 * intersection /
            (pred_mask.sum() + actual_mask.sum() + 1e-8)
        )

        iou = intersection / (union + 1e-8)

        results.append({
            "Image": names[i],
            "Dice": dice * 100,
            "IoU": iou * 100
        })

for filename in image_files:
    stem = os.path.splitext(filename)[0]

    if stem not in mask_files:
        print("Missing mask:", filename)
        continue

    image = Image.open(
        os.path.join(IMAGE_DIR, filename)
    ).convert("RGB").resize((256, 256))

    mask = Image.open(
        os.path.join(MASK_DIR, mask_files[stem])
    ).convert("L").resize(
        (256, 256), Image.Resampling.NEAREST
    )

    batch_images.append(
        np.array(image, dtype=np.float32) / 255.0
    )
    batch_masks.append(np.array(mask))
    batch_names.append(filename)

    if len(batch_images) == batch_size:
        evaluate_batch(batch_images, batch_masks, batch_names)
        batch_images, batch_masks, batch_names = [], [], []

if batch_images:
    evaluate_batch(batch_images, batch_masks, batch_names)

df = pd.DataFrame(results)
df.to_csv("/content/InfraScan_Final_UNet_Results.csv", index=False)

print("\n========== FINAL U-NET RESULTS ==========")
print("Images evaluated:", len(df))
print("Average Dice:", round(df["Dice"].mean(), 2), "%")
print("Average IoU:", round(df["IoU"].mean(), 2), "%")
print("\nCSV saved successfully!")
```
