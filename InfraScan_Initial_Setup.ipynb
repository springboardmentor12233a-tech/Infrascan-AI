```python

import os
from PIL import Image
from collections import Counter

base_folder = "/content/InfraScan_Dataset"
categories = ["cracked", "uncracked"]

print("INFRASCAN - DATASET SUMMARY")
print("-" * 35)

total_images = 0

for category in categories:
    folder = os.path.join(base_folder, category)
    image_files = [
        f for f in os.listdir(folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    sizes = Counter()

    for filename in image_files:
        path = os.path.join(folder, filename)
        with Image.open(path) as img:
            sizes[img.size] += 1

    count = len(image_files)
    total_images += count

    print(f"{category.title()} images: {count}")
    print(f"Image dimensions: {dict(sizes)}")
    print()

print("Total images:", total_images)
print("Dataset summary completed!")
```

```python
import os
import shutil
import random

source = "/content/InfraScan_Dataset"
output = "/content/InfraScan_Split"

random.seed(42)

for split in ["train", "validation"]:
    for category in ["cracked", "uncracked"]:
        os.makedirs(
            os.path.join(output, split, category),
            exist_ok=True
        )

for category in ["cracked", "uncracked"]:
    files = [
        f for f in os.listdir(os.path.join(source, category))
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(files)
    split_index = max(1, int(len(files) * 0.8))

    for filename in files[:split_index]:
        shutil.copy2(
            os.path.join(source, category, filename),
            os.path.join(output, "train", category, filename)
        )

    for filename in files[split_index:]:
        shutil.copy2(
            os.path.join(source, category, filename),
            os.path.join(output, "validation", category, filename)
        )

print("Sample dataset split completed!")

for split in ["train", "validation"]:
    for category in ["cracked", "uncracked"]:
        folder = os.path.join(output, split, category)
        print(f"{split.title()} - {category}: {len(os.listdir(folder))}")
```

```python
import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(256, 256, 3)),
    tf.keras.layers.Rescaling(1.0 / 255),

    tf.keras.layers.Conv2D(16, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import tensorflow as tf

train_dataset = tf.keras.utils.image_dataset_from_directory(
    "/content/InfraScan_Split/train",
    image_size=(256, 256),
    batch_size=4,
    label_mode="binary",
    shuffle=True,
    seed=42
)

validation_dataset = tf.keras.utils.image_dataset_from_directory(
    "/content/InfraScan_Split/validation",
    image_size=(256, 256),
    batch_size=4,
    label_mode="binary",
    shuffle=False
)

print("Training and validation datasets loaded!")
print("Classes:", train_dataset.class_names)
```

```python
history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=5
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))

plt.plot(history.history["accuracy"], marker="o", label="Training Accuracy")
plt.plot(history.history["val_accuracy"], marker="o", label="Validation Accuracy")

plt.title("InfraScan CNN Training Progress")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.xticks(range(5), range(1, 6))
plt.ylim(0, 1)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "/content/InfraScan_Training_Accuracy.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

print("Training accuracy graph saved!")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred = []

for images, labels in validation_dataset:
    scores = model.predict(images, verbose=0).flatten()
    predictions = (scores < 0.5).astype(int)

    y_true.extend(labels.numpy().flatten().astype(int))
    y_pred.extend(predictions)

print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred, labels=[0, 1]))

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=validation_dataset.class_names,
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=validation_dataset.class_names,
    zero_division=0
)

matrix = confusion_matrix(y_true, y_pred, labels=[0, 1])

with open("/content/InfraScan_Evaluation_Report.txt", "w") as f:
    f.write("INFRASCAN - PRELIMINARY MODEL EVALUATION\n\n")
    f.write("Dataset: 14-image test sample\n")
    f.write("Validation images: 3\n\n")
    f.write("Confusion Matrix:\n")
    f.write(str(matrix))
    f.write("\n\nClassification Report:\n")
    f.write(report)
    f.write("\n\nNote: Results are preliminary and not reliable\n")
    f.write("for real-world crack detection.\n")

print("Evaluation report saved successfully!")
```

```python
import os
import numpy as np
from PIL import Image

root = "/content/InfraScan_Preprocessed"
total = 0
errors = []

for category in sorted(os.listdir(root)):
    folder = os.path.join(root, category)
    if not os.path.isdir(folder):
        continue

    count = 0
    for filename in os.listdir(folder):
        path = os.path.join(folder, filename)
        try:
            with Image.open(path) as img:
                img = img.convert("RGB").resize((256, 256))
                pixels = np.asarray(img, dtype=np.float32) / 255.0
                assert pixels.shape == (256, 256, 3)
                count += 1
        except Exception as e:
            errors.append((filename, str(e)))

    print(f"{category}: {count} images verified and preprocessed")
    total += count

print("\nTotal images:", total)
print("Errors:", len(errors))
print("Pixel normalization range: 0 to 1")
print("Preprocessing verification completed.")
```

```python
import tensorflow as tf

data_dir = "/content/InfraScan_Preprocessed"

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(256, 256),
    batch_size=16
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(256, 256),
    batch_size=16
)

class_names = train_ds.class_names
print("Classes:", class_names)

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(256, 256, 3)),
    tf.keras.layers.Rescaling(1./255),
    tf.keras.layers.Conv2D(16, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(32, 3, activation="relu"),
    tf.keras.layers.MaxPooling2D(),
    tf.keras.layers.Conv2D(64, 3, activation="relu"),
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(len(class_names), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)

model.save("/content/InfraScan_6Class_CNN.keras")
print("Six-class CNN trained and saved.")
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report

y_true = []
y_pred = []

for images, labels in val_ds:
    predictions = model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(predictions, axis=1))

print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=list(range(len(class_names))),
    target_names=class_names,
    zero_division=0
))
```

```python

# ============================================================
# INFRASCAN — COMPLETE DEMO APPLICATION
# Login | Role Access | Asset Management | Image Inspection
# ============================================================

import os
import sys
import sqlite3
import hashlib
import subprocess
import importlib.util

# 1. Install Gradio if missing
if importlib.util.find_spec("gradio") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "gradio"
    ])

import gradio as gr
import numpy as np
from PIL import Image

DB_PATH = "/content/InfraScan.db"
MODEL_PATH = "/content/InfraScan_6Class_CNN.keras"

# 2. Database initialization
def connect_db():
    return sqlite3.connect(DB_PATH, timeout=30)

def hash_password(password):
    return hashlib.sha256(password.encode("utf-8")).hexdigest()

with connect_db() as conn:
    conn.execute("""
        CREATE TABLE IF NOT EXISTS users (
            username TEXT PRIMARY KEY,
            password_hash TEXT NOT NULL,
            role TEXT NOT NULL
        )
    """)
    conn.execute("""
        CREATE TABLE IF NOT EXISTS assets (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            asset_name TEXT NOT NULL,
            asset_type TEXT NOT NULL,
            location TEXT,
            status TEXT DEFAULT 'Active'
        )
    """)

    conn.execute(
        "INSERT OR IGNORE INTO users VALUES (?, ?, ?)",
        ("admin", hash_password("Admin@123"), "admin")
    )
    conn.execute(
        "INSERT OR IGNORE INTO users VALUES (?, ?, ?)",
        ("user", hash_password("User@123"), "user")
    )

    conn.execute("""
        INSERT INTO assets (asset_name, asset_type, location, status)
        SELECT 'Bridge A', 'Bridge', 'Mysore', 'Active'
        WHERE NOT EXISTS (
            SELECT 1 FROM assets WHERE asset_name = 'Bridge A'
        )
    """)

print("Database ready:", DB_PATH)

# 3. Load optional CNN model
model = None
class_names = [
    "deck_cracked",
    "deck_uncracked",
    "pavement_cracked",
    "pavement_uncracked",
    "wall_cracked",
    "wall_uncracked"
]

if os.path.isfile(MODEL_PATH):
    try:
        import tensorflow as tf
        model = tf.keras.models.load_model(MODEL_PATH)
        print("CNN model loaded successfully.")
    except Exception as e:
        print("CNN model unavailable:", str(e))
else:
    print("CNN model not found; image preprocessing remains available.")

# 4. Login
def login(username, password):
    username = (username or "").strip()

    with connect_db() as conn:
        row = conn.execute(
            "SELECT password_hash, role FROM users WHERE username = ?",
            (username,)
        ).fetchone()

    if row and row[0] == hash_password(password or ""):
        role = row[1]
        return (
            f"Login successful: {username} ({role})",
            role,
            gr.update(visible=True),
            gr.update(visible=(role == "admin"))
        )

    return (
        "Invalid username or password.",
        None,
        gr.update(visible=False),
        gr.update(visible=False)
    )

def logout():
    return (
        None,
        "Logged out.",
        gr.update(visible=False),
        gr.update(visible=False)
    )

# 5. Register infrastructure asset
def register_asset(name, asset_type, location, role):
    if role != "admin":
        return "Access denied. Admin login required."

    name = (name or "").strip()
    location = (location or "").strip()

    if not name or not asset_type:
        return "Please enter the asset name and type."

    try:
        with connect_db() as conn:
            conn.execute("""
                INSERT INTO assets
                    (asset_name, asset_type, location, status)
                VALUES (?, ?, ?, 'Active')
            """, (name, asset_type, location))

        return f"Asset '{name}' registered successfully."

    except Exception as e:
        return f"Asset registration failed: {e}"

# 6. View assets — returns plain text
def view_assets(role):
    if role not in ("admin", "user"):
        return "Please log in first."

    try:
        with connect_db() as conn:
            rows = conn.execute("""
                SELECT id, asset_name, asset_type, location, status
                FROM assets
                ORDER BY id DESC
            """).fetchall()

        if not rows:
            return "No assets registered yet."

        lines = [
            "ID | Asset Name | Type | Location | Status",
            "-" * 58
        ]

        for row in rows:
            lines.append(
                f"{row[0]} | {row[1]} | {row[2]} | "
                f"{row[3] or '-'} | {row[4]}"
            )

        return "\n".join(lines)

    except Exception as e:
        return f"Could not load assets: {e}"

# 7. Image preprocessing and experimental prediction
def inspect_image(image, role):
    if role not in ("admin", "user"):
        return None, "Please log in first."

    if image is None:
        return None, "Please upload a concrete image."

    try:
        img = Image.fromarray(np.asarray(image)).convert("RGB")
        img = img.resize((256, 256))
        arr = np.asarray(img, dtype=np.float32) / 255.0

        if model is None:
            return img, (
                "Preprocessing successful.\n"
                "Image size: 256 x 256\n"
                "Color: RGB\n"
                "Normalized pixel range: 0–1\n"
                "Prediction unavailable because the CNN model is not loaded."
            )

        prediction = np.asarray(
            model.predict(np.expand_dims(arr, axis=0), verbose=0)
        ).reshape(-1)

        if len(prediction) != len(class_names):
            return img, (
                f"Model returned {len(prediction)} output values. "
                "Expected six classes; check the loaded model."
            )

        idx = int(np.argmax(prediction))
        label = class_names[idx]
        confidence = float(prediction[idx]) * 100

        return img, (
            f"Predicted category: {label}\n"
            f"Model score: {confidence:.2f}%\n\n"
            "Research/demo result only. This experimental model is not "
            "reliable enough for real infrastructure safety decisions."
        )

    except Exception as e:
        return None, f"Image inspection failed: {e}"

# 8. Build Gradio interface
with gr.Blocks(title="InfraScan") as demo:
    gr.Markdown(
        "# InfraScan\n"
        "### AI-Powered Crack & Damage Assessment for Civil Infrastructure"
    )

    role_state = gr.State(value=None)

    with gr.Group():
        gr.Markdown("### Login")
        username_box = gr.Textbox(label="Username", value="admin")
        password_box = gr.Textbox(
            label="Password", type="password", value="Admin@123"
        )
        login_button = gr.Button("Login", variant="primary")
        login_status = gr.Textbox(label="Login status", interactive=False)
        logout_button = gr.Button("Logout")

    with gr.Group(visible=False) as main_area:
        with gr.Tabs():
            with gr.Tab("Inspect Image"):
                gr.Markdown("Upload a concrete image for analysis.")
                image_input = gr.Image(
                    type="numpy", label="Upload concrete image"
                )
                inspect_button = gr.Button("Inspect Image")
                processed_output = gr.Image(label="Processed image")
                prediction_output = gr.Textbox(
                    label="Inspection result", lines=5
                )

            with gr.Tab("Register Asset", visible=False) as admin_tab:
                gr.Markdown("Admin-only asset registration")
                asset_name_box = gr.Textbox(label="Asset name")
                asset_type_box = gr.Dropdown(
                    choices=["Bridge", "Road", "Building", "Wall", "Other"],
                    label="Asset type",
                    value="Road"
                )
                asset_location_box = gr.Textbox(label="Location")
                register_button = gr.Button("Register Asset")
                register_status = gr.Textbox(
                    label="Registration status", interactive=False
                )

            with gr.Tab("View Assets"):
                refresh_button = gr.Button("Refresh asset list")
                asset_list = gr.Textbox(
                    label="Registered infrastructure assets",
                    lines=12,
                    interactive=False
                )

    login_button.click(
        fn=login,
        inputs=[username_box, password_box],
        outputs=[login_status, role_state, main_area, admin_tab]
    )

    logout_button.click(
        fn=logout,
        inputs=[],
        outputs=[role_state, login_status, main_area, admin_tab]
    )

    register_button.click(
        fn=register_asset,
        inputs=[
            asset_name_box, asset_type_box,
            asset_location_box, role_state
        ],
        outputs=[register_status]
    )

    refresh_button.click(
        fn=view_assets,
        inputs=[role_state],
        outputs=[asset_list]
    )

    inspect_button.click(
        fn=inspect_image,
        inputs=[image_input, role_state],
        outputs=[processed_output, prediction_output]
    )

print("Starting InfraScan demo...")
demo.launch(share=True)
```
