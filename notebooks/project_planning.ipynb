# Milestone 1 Project Planning

## InfraScan – AI-Powered Crack & Damage Assessment for Civil Infrastructure

This notebook documents the initial planning for InfraScan. It describes the intended problem, users, workflow, requirements, boundaries, and development milestones. It is a planning document: items described as future work are not claimed to be implemented.

## 1. Problem statement

Manual inspection of concrete infrastructure is important but can be difficult, slow, and inconsistent. Inspectors may need to examine large structures, work at height, or access locations that are remote, poorly lit, or difficult to reach. Recording crack observations by hand can also make it harder to compare inspections over time and to maintain consistent documentation between inspection teams.

A crack can be small, low-contrast, partially hidden, or visually similar to surface texture, stains, joints, and shadows. Human judgment remains essential, but reviewing many images manually can increase workload and may lead to differences in how damage is identified or recorded. InfraScan is planned as a decision-support system that can help organize image-based assessment and highlight areas that require expert attention. It is not intended to replace qualified engineering judgment or formal safety inspections.

## 2. Project objectives

The project objectives for the planned system are:

1. Build a clear workflow for uploading and validating infrastructure images.
2. Prepare images consistently through documented preprocessing steps.
3. Develop a crack classification model that distinguishes images with and without visible cracks.
4. Investigate crack detection methods that can locate likely damaged regions.
5. Investigate segmentation methods that can outline crack pixels or areas more precisely.
6. Present predictions and supporting visual results in an understandable format.
7. Produce a report that inspectors and engineers can review and use as an aid to their existing process.
8. Measure model performance with suitable validation data and metrics rather than relying only on visual examples.
9. Record limitations, uncertainty, and cases that require human review.

The first milestone focuses on planning, dataset understanding, and preparation. The objectives above describe the overall direction and do not mean that every capability is currently available.

## 3. Target infrastructure

The intended application area is concrete civil infrastructure, including:

- Concrete residential, commercial, and public buildings.
- Concrete bridges, bridge decks, piers, and related components.
- Concrete roads, pavements, parking areas, and other transport surfaces.
- Tunnels, retaining walls, dams, foundations, culverts, and other concrete structures.

The appearance of concrete varies with lighting, weathering, construction methods, surface finish, camera distance, and repair history. These variations must be considered when collecting data and evaluating any future model.

## 4. Target users

The primary target users are:

- **Infrastructure inspectors:** to review image evidence and prioritize areas for closer inspection.
- **Civil engineers:** to use organized visual evidence as one input to engineering assessment and maintenance planning.
- **Maintenance teams:** to record suspected damage, compare observations, and help plan follow-up work.

The system should use clear language and visual explanations because users may have different levels of experience with machine-learning tools. Final decisions about structural safety, repair urgency, and engineering action must remain with appropriately qualified professionals.

## 5. Planned project workflow

The intended end-to-end workflow is:

**Image upload → Image validation → Preprocessing → Crack classification → Crack detection → Segmentation → Results and reporting**

### Stage descriptions

1. **Image upload:** Accept an image selected by the user or provided by a future inspection data source.
2. **Image validation:** Check that the file is a supported image, can be read, and contains usable image data.
3. **Preprocessing:** Apply documented operations such as resizing, normalization, and format conversion when needed. The original image should be preserved for reference.
4. **Crack classification:** Predict whether a visible crack is present in the image or selected image region.
5. **Crack detection:** Locate likely crack regions using bounding boxes, heatmaps, or another suitable detection representation.
6. **Segmentation:** Produce a more detailed pixel-level or region-level outline when the selected model and training data support it.
7. **Results and reporting:** Show predictions, confidence information, visual evidence, warnings, and an exportable summary for review.

During Milestone 1, this workflow is a design plan. The dataset analysis work supports preparation for later stages; it does not by itself provide detection or segmentation.

## 6. Functional requirements

Functional requirements describe what the planned system should do:

1. Allow a user to select or upload a supported infrastructure image.
2. Validate file type, readability, image dimensions, and other basic input conditions.
3. Report a useful error when an image is missing, corrupted, unsupported, or unsuitable for processing.
4. Apply the same documented preprocessing steps used during model development.
5. Provide a crack/no-crack classification result when a trained classification model is available.
6. Display a confidence or uncertainty indication and avoid presenting a prediction as a guaranteed fact.
7. Support a future detection result that identifies likely crack locations.
8. Support a future segmentation result that shows the estimated crack area or outline when appropriate data and a validated model are available.
9. Show the original image alongside relevant visual results where possible.
10. Generate a readable results summary containing the input reference, analysis date, model result, and warnings.
11. Allow an authorized user to review results before using them in maintenance or engineering decisions.

At the planning stage, these are requirements, not claims that all functions are already implemented.

## 7. Non-functional requirements

Non-functional requirements describe quality and operational expectations:

- **Usability:** The interface and reports should be understandable to inspectors and engineers without requiring machine-learning expertise.
- **Reliability:** Invalid inputs and processing failures should be reported clearly rather than producing misleading success messages.
- **Performance:** Typical images should be processed within a practical time for an inspection workflow; the target response time should be measured during development.
- **Reproducibility:** Dataset versions, preprocessing settings, model versions, and evaluation results should be recorded.
- **Maintainability:** Components should be organized so that models, preprocessing, and reporting can be improved independently.
- **Scalability:** The design should leave room for additional infrastructure types, image batches, and future model versions.
- **Security and privacy:** Uploaded images and reports should be protected from unauthorized access, and sensitive project information should not be exposed unnecessarily.
- **Explainability:** Results should include appropriate visual evidence and limitations so that users can review them critically.
- **Safety:** The system must clearly state that automated output is decision support and is not a substitute for a qualified structural assessment.

## 8. Project scope and limitations

### In scope

- Analysis of concrete surface images for visible crack-related patterns.
- A binary crack classification baseline using a labeled image dataset.
- Investigation of localization and segmentation as later development stages.
- Image validation, documented preprocessing, evaluation, and clear reporting.
- A workflow that assists human inspection rather than replacing it.

### Limitations and out-of-scope uses

- An image model cannot determine structural safety or the underlying cause of damage by itself.
- A visible crack prediction does not establish crack depth, load capacity, repair priority, or remaining service life.
- Performance may decrease for lighting, surfaces, camera angles, materials, or infrastructure types not represented in the training data.
- Classification, detection, and segmentation require suitable labeled data and separate evaluation; a classification dataset alone does not automatically support accurate localization or pixel-level segmentation.
- The planned system should not be used as the sole basis for emergency response, structural certification, or maintenance approval.
- Privacy, connectivity, hardware, and deployment constraints will need further definition before production use.

The current project materials support dataset exploration and planning. The user-facing workflow and advanced model stages remain future development unless separately implemented and validated.

## 9. Milestone-wise development plan

### Milestone 1 — Planning and dataset understanding

- Define the problem, users, workflow, requirements, scope, and limitations.
- Inspect the available dataset structure, class counts, sample images, image properties, and readability.
- Record assumptions and identify gaps in labels, coverage, and evaluation data.

### Milestone 2 — Data preparation and baseline classification

- Define reproducible train, validation, and test splits without unintended leakage.
- Implement preprocessing and establish a baseline classification model.
- Evaluate accuracy alongside class-sensitive metrics such as precision, recall, F1-score, and a confusion matrix.

### Milestone 3 — Classification evaluation and improvement

- Review errors and difficult examples.
- Compare appropriate model approaches and training settings.
- Check robustness across relevant image conditions and document the selected model and limitations.

### Milestone 4 — Crack detection

- Collect or prepare bounding-box or region-level annotations if detection is required.
- Train and evaluate a detection approach using suitable localization metrics.
- Add visual overlays for human review.

### Milestone 5 — Crack segmentation

- Obtain or create reliable pixel-level masks where segmentation is justified.
- Train and evaluate a segmentation model using appropriate overlap and boundary metrics.
- Compare segmentation outputs with expert review and document known failure cases.

### Milestone 6 — Application and reporting workflow

- Implement input validation, user-facing processing, result visualization, and report generation.
- Add clear uncertainty messages, logging, and error handling.
- Test the complete planned workflow with representative examples.

### Milestone 7 — Validation and deployment readiness

- Conduct broader testing with representative infrastructure images and qualified reviewers.
- Review privacy, security, performance, maintainability, and monitoring requirements.
- Prepare documentation and a controlled release plan only if the evidence supports the intended use.

These milestones are a proposed sequence. Completion of a milestone should be supported by measurable results and documentation, not only by the presence of code.

## Current capabilities versus future development

At the start of Milestone 1, the project has planning documentation and dataset-analysis work. The classification, detection, segmentation, user interface, reporting, and production deployment stages described above are planned capabilities. They should only be described as implemented after they have been built, tested, and evaluated with appropriate evidence.
