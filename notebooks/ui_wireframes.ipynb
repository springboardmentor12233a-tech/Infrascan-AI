# InfraScan UI Wireframes

This notebook documents a planned user interface and user experience for **InfraScan - AI-Powered Crack & Damage Assessment for Civil Infrastructure**.

These wireframes are design specifications, not implemented screens. They describe a beginner-friendly web application that supports image-based inspection while keeping qualified professionals responsible for engineering decisions.

## 1. UI design overview and design goals

### Purpose

Provide a clear, low-complexity interface for uploading infrastructure images, organizing inspections, reviewing planned AI assistance, and creating reports.

### Design goals

- Make the next action obvious for first-time users.
- Keep original images, predictions, confidence information, and warnings visible together.
- Use plain language instead of unexplained machine-learning terminology.
- Prevent accidental loss of uploaded evidence or inspection notes.
- Clearly distinguish automated assistance from confirmed engineering conclusions.
- Support keyboard use, readable contrast, responsive layouts, and accessible error messages.

The current project contains planning and dataset-analysis documentation. The screens described here are planned and are not claimed to be implemented.

## 2. User roles and interface permissions

| Role | Planned permissions |
| --- | --- |
| Inspector | Create inspections, manage assigned assets, upload images, request analysis, review results, and create reports. |
| Civil engineer | Review assigned inspections and AI results, add technical notes, approve or correct review status, and download reports. |
| Maintenance team member | View assigned assets and results, add follow-up status or maintenance notes, and download permitted reports. |
| Administrator | Manage users and roles, view system-level activity, configure permitted values, and access audit information. |

The frontend should hide actions a role cannot use, but the backend must also enforce every permission. Role permissions shown here are planned behavior, not an implemented access-control system.

## 3. Login and registration screen

### Purpose
Allow an authorized user to sign in securely and, if enabled by an administrator, request or complete account registration.

### Main UI components
- InfraScan logo and short product description.
- Email field and password field.
- Optional remember-session control.
- Registration link or invitation-code flow.
- Forgot-password link and accessible error area.

### Important buttons and actions
- **Sign in:** Validate fields and submit credentials.
- **Create account:** Open the planned registration flow.
- **Reset password:** Begin account recovery.
- **Show or hide password:** Change visibility without exposing it by default.

### Information displayed
- Required-field guidance and password requirements.
- Specific, safe error messages such as invalid credentials or unavailable service.
- A reminder that access is restricted to authorized project users.

```text
+------------------------------------------------+
| InfraScan                                      |
| Crack and damage assessment support            |
|                                                |
| Email        [____________________________]    |
| Password     [____________________________]    |
| [ ] Remember this device   [Show password]     |
|                                                |
|              [ Sign in ]                      |
| Forgot password?       Request an account      |
| [Accessible error or status message]           |
+------------------------------------------------+
```

## 4. Main dashboard

### Purpose
Give users a quick overview of their permitted infrastructure work and provide clear entry points to start or continue an inspection.

### Main UI components
- Persistent header with product name, user menu, and sign-out action.
- Sidebar or top navigation.
- Summary cards for assigned assets, active inspections, completed inspections, and items needing review.
- Recent inspections table with asset, date, status, and reviewer action.
- Primary call-to-action for creating an inspection.

### Important buttons and actions
- **New inspection:** Open the inspection creation screen.
- **View all inspections:** Open filtered inspection history.
- **Open:** View a selected inspection.
- **Refresh:** Request current data from the backend.

### Information displayed
- Counts should be scoped to the signed-in user's permissions.
- Status labels should use both text and color, not color alone.
- Empty states should explain how to create the first asset or inspection.

```text
[InfraScan]  Dashboard  Assets  Inspections  Reports       [User] [Sign out]
-----------------------------------------------------------------------
| Assets: 24 | Active inspections: 3 | Awaiting review: 5 | Done: 61 |
-----------------------------------------------------------------------
Recent inspections                         [ + New inspection ]
| Asset              Date          Status          Action             |
| Riverside Bridge   2026-10-30    Awaiting review [Open]             |
| Civic Building     2026-10-28    Completed       [Open]             |
```


## 5. Infrastructure asset management screen

### Purpose
Create, review, search, and update the concrete buildings, bridges, roads, and other structures that belong to the user's permitted projects.

### Main UI components
- Search field and infrastructure-type filter.
- Asset table or cards with name, type, location, last inspection, and status.
- Asset detail panel with description and inspection history.
- Form fields for asset name, type, location, and description.

### Important buttons and actions
- **Add asset:** Open a validated asset form.
- **Open asset:** View details and related inspections.
- **Edit:** Update permitted asset metadata.
- **Archive:** Remove an asset from active lists without deleting inspection evidence.
- **Start inspection:** Begin an inspection for the selected asset.

### Information displayed
- Asset identifier, infrastructure type, location, description, creation date, and last inspection status.
- Clear confirmation before archiving.
- Validation feedback for missing names, invalid types, or incomplete locations.

## 6. Image upload screen

### Purpose
Let a user add image evidence to an inspection while validating the file before analysis.

### Main UI components
- Drag-and-drop area and file picker.
- Image preview with filename and basic metadata.
- Upload progress indicator.
- Validation checklist for supported format, file size, readability, and dimensions.
- List of selected images with remove controls.

### Important buttons and actions
- **Choose images:** Open the local file picker.
- **Remove:** Remove a selected file before submission.
- **Upload images:** Transfer validated files.
- **Continue to assessment:** Start the planned analysis request after upload succeeds.
- **Cancel:** Leave without saving, with confirmation if files have been selected.

### Information displayed
- Supported formats and maximum file size.
- Original filename, detected format, dimensions, and upload status.
- Explicit errors for corrupted or unsupported files.
- A note that the original image is preserved as inspection evidence.

```text
Upload images for: Riverside Bridge / Inspection 2026-10-30
+---------------------------------------------------------------+
| Drag images here or [Choose images]                          |
| JPG, PNG | Maximum size: configured by deployment              |
+---------------------------------------------------------------+
| [thumbnail] bridge_01.jpg  1920x1080  Valid     [Remove]       |
| [thumbnail] bridge_02.jpg  1920x1080  Checking...             |
                                                               |
| [Cancel]                         [Upload images]              |
```

## 7. Inspection creation screen

### Purpose
Create a structured inspection record before adding image evidence and requesting analysis.

### Main UI components
- Asset selector with search.
- Inspection date and optional site-reference fields.
- Inspector and assigned reviewer fields based on permissions.
- Notes and environmental context fields.
- Step indicator: Details → Images → Assessment → Review.

### Important buttons and actions
- **Save draft:** Store incomplete information for later.
- **Continue:** Validate details and open image upload.
- **Cancel:** Return to the previous screen.
- **Discard draft:** Require confirmation before removing an unsent draft.

### Information displayed
- Selected asset name, type, and location.
- Inspection date, current user, status, and notes.
- Required fields and validation errors next to the relevant field.
- A warning that AI output is decision support, not a safety certification.

## 8. AI assessment results screen

### Purpose
Present planned AI results alongside the original image so a human reviewer can assess the evidence and uncertainty.

### Main UI components
- Original image viewer with zoom and fit controls.
- Annotated-image placeholder for future bounding boxes, heatmaps, or segmentation masks.
- Result summary card with defect class and confidence score.
- Model name, model version, analysis time, and processing status.
- Review notes and reviewer-status controls.
- Clear warning banner about automated limitations.

### Important buttons and actions
- **Run assessment:** Request the selected available model.
- **View annotation:** Toggle planned overlays.
- **Accept result / Mark for review / Correct result:** Record a qualified user's review state.
- **Add note:** Add contextual observations.
- **Generate report:** Move to report generation after review.

### Information displayed
- Predicted defect class, for example `Crack` or `No visible crack`.
- Confidence score with a plain-language explanation that it is not certainty.
- Annotated-image placeholder when detection or segmentation is not yet available.
- Failure, pending, and unavailable-model states that do not look like successful results.

```text
Assessment result: bridge_01.jpg       [Pending review]
+--------------------------+-------------------------------+
| Original image           | Annotation placeholder       |
|                          | [Boxes / mask / heatmap]     |
|                          |                               |
+--------------------------+-------------------------------+
| Predicted class: Crack                                  |
| Confidence: 0.87 (model score, not a guarantee)         |
| Model: planned-classifier v0.1                          |
| [Accept] [Mark for review] [Add note] [Generate report] |
+---------------------------------------------------------+
```

Classification may be developed before detection or segmentation. The overlay area must show an explicit unavailable state until those future capabilities are actually implemented and validated.

## 9. Inspection history screen

### Purpose
Help users find previous inspections and compare status, dates, assets, and review outcomes.

### Main UI components
- Search by asset name or inspection identifier.
- Filters for date range, asset type, status, reviewer status, and assigned user.
- Sortable table with pagination.
- Optional compact timeline for one asset.
- Export or open action according to permissions.

### Important buttons and actions
- **Apply filters / Clear filters:** Refine or reset the list.
- **Open inspection:** View images, results, notes, and reports.
- **Compare inspections:** Planned comparison for two permitted inspections.
- **Create report:** Open a report request for a selected inspection.

### Information displayed
- Asset, inspection date, inspector, image count, processing status, review status, and last update.
- Empty and no-result states with guidance.
- Pagination information so users understand the result set.

## 10. Report generation and download screen

### Purpose
Allow a permitted user to choose the reviewed content and create a traceable inspection summary.

### Main UI components
- Inspection summary and report title field.
- Selectable sections: asset details, images, reviewed AI results, notes, and limitations.
- Report preview or generation status.
- Existing report versions and creation dates.

### Important buttons and actions
- **Generate report:** Start report creation after validation.
- **Preview:** Show a non-final preview where supported.
- **Download:** Download a completed report the user may access.
- **Regenerate version:** Create a new version while retaining history.

### Information displayed
- Report status: generating, ready, failed, or archived.
- Model versions and assessment timestamps included in the report.
- Warnings that AI findings require professional review.
- Clear error messages when an image or result is unavailable.

### Purpose
Let users review account information and manage settings they are authorized to change.

### Main UI components
- Profile form with name and email.
- Role display that cannot be changed by ordinary users.
- Password-change or identity-provider link.
- Notification preferences.
- Accessibility and display preferences.
- Administrator-only user and role management area.

### Important buttons and actions
- **Save changes:** Validate and update editable profile data.
- **Change password:** Start a secure password-change flow where applicable.
- **Sign out all sessions:** Revoke other sessions if supported.

### Information displayed
- Account name, normalized email, role, account status, and last update.

## 12. Navigation structure and user journey

### Planned navigation

```mermaid
flowchart TD
    Login[Login / Registration] --> Dashboard[Dashboard]
    Dashboard --> Assets[Infrastructure Assets]
    Dashboard --> History[Inspection History]
    Dashboard --> NewInspection[Create Inspection]
    Assets --> AssetDetail[Asset Detail]
    AssetDetail --> NewInspection
    NewInspection --> Upload[Upload Images]
    Upload --> Results[AI Assessment Results]
    Results --> History
    Results --> Reports[Reports]
    History --> Results
    Reports --> History
    Dashboard --> Profile[Profile and Settings]
```

### Main user journey

1. Sign in.
2. Choose an existing asset or create one if permitted.
3. Create an inspection draft and enter context.
4. Upload images and resolve validation errors.
5. Request the available planned assessment.
6. Review class, confidence, image evidence, and any annotation placeholder.
7. Add professional notes or mark the result for further review.
8. Generate and download a report when permitted.
9. Return to history to track the inspection over time.

The application should preserve drafts and display explicit progress so users do not lose their place.

## 13. Responsive design and accessibility considerations

### Responsive design

- Use a mobile-first layout that works on phones used at inspection sites, tablets, and desktop monitors.
- Collapse the sidebar into a labeled menu on narrow screens.
- Keep image viewers usable with pinch or keyboard zoom and provide fit-to-screen behavior.
- Turn wide tables into cards or horizontally scrollable regions without hiding essential information.
- Keep primary actions visible while scrolling through long forms.

### Accessibility

- Use semantic headings, labels, landmarks, and predictable tab order.
- Provide keyboard access for upload, dialogs, image controls, tables, and navigation.
- Use sufficient color contrast and pair status colors with text or icons.
- Give images meaningful alternative text; for technical images, include asset and inspection context.
- Announce upload progress, validation results, processing changes, and errors to assistive technology.
- Do not rely on hover-only information or color alone to communicate crack status.
- Respect reduced-motion and text-size preferences where practical.

These are planned design requirements. They should be checked with automated accessibility tools and manual keyboard and screen-reader testing after implementation.
